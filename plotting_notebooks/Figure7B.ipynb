```python
import pandas as pd
import plotnine as p9
from plotnine_prism import *
```

```python
df = pd.read_csv("../plotting_notebooks/resources/train_size.csv")
df["Model"] = pd.Categorical(df.model, ["DeepCell", "STNet", "BLEEP"])
df.n_slides = df.n_slides.astype("category")
df
```

```python
g = (p9.ggplot(df, p9.aes("n_slides", "pearson_mean", color="Model")) 
 + p9.geom_point()
 + p9.geom_line(p9.aes(group="model"), linetype="dashed")
 + p9.geom_errorbar(p9.aes(ymin="pearson_mean-pearson_std", ymax="pearson_mean+pearson_std"), width=0.2)
 + p9.theme_bw()
 + p9.ylab("Pearson correlation (n_genes=100)")
 + p9.xlab("Training patients")
 + p9.theme(figure_size=(7, 6))
# + p9.ggtitle("Lung Xenium HEST1K (n=20)")
 + scale_color_prism(palette = "colors")
 + p9.theme(panel_spacing_y=0, panel_spacing_x=0, 
            #axis_text_x = p9.element_blank(), 
            legend_position="right",
            text=p9.element_text(size=17),
            strip_text=p9.element_text(size=17),
            legend_title=p9.element_text(size=17),
            legend_text=p9.element_text(size=16))

)
g.save("figures/Figure7B_train_size.png", dpi=300)
g
```
